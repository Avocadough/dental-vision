```python
import json

# โหลดไฟล์ JSON
with open("/content/drive/MyDrive/teeth/keypoint_labelV2.json", "r", encoding="utf-8") as file:
    data = json.load(file)

# ตรวจสอบโครงสร้าง JSON
images = []
annotations = []

for item in data:
    image_info = {
        "id": item["id"],
        "file_name": item["data"]["img"].split("/")[-1],
        "width": item["annotations"][0]["result"][0]["original_width"],
        "height": item["annotations"][0]["result"][0]["original_height"]
    }
    images.append(image_info)

    for annotation in item["annotations"][0]["result"]:
        keypoint = annotation["value"]
        annotation_info = {
            "image_id": item["id"],
            "category_id": 5,  # หมวดหมู่ (แก้ไขตามหมวดของคุณ)
            "keypoints": [
                keypoint["x"], keypoint["y"], 2  # 2 คือ visibility
            ],
            "num_keypoints": 24
        }
        annotations.append(annotation_info)

# สร้าง JSON ใหม่ใน COCO format
coco_data = {
    "images": images,
    "annotations": annotations,
    "categories": [{"id": 1, "name": "keypoint", "keypoints": ["kp1"], "skeleton": []}]
}

# บันทึกไฟล์ JSON ที่แปลงแล้ว
with open("coco_keypoints.json", "w", encoding="utf-8") as outfile:
    json.dump(coco_data, outfile, indent=4)
```
